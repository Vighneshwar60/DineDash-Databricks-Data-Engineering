```python
%sql
WITH CustomerRestaurantCounts AS(
SELECT c.customer_id,c.name AS customer_name,r.restaurant_id,r.name AS restaurant_name,COUNT(o.order_id) AS order_count,
ROW_NUMBER() OVER(
    PARTITION BY c.customer_id
    ORDER BY COUNT(o.order_id) DESC, r.restaurant_id ASC

) as rn

FROM customers_bronze c
JOIN orders_bronze o ON c.customer_id = o.customer_id
JOIN resaurants_filtered r ON o.restaurant_id = r.restaurant_id
GROUP BY c.customer_id, c.name, r.restaurant_id, r.name 

)

SELECT customer_id,customer_name,restaurant_name,order_count 
FROM CustomerRestaurantCounts
WHERE rn = 1;
```

q2

```python
%sql
With monthly_orders AS(
    SELECT
        DATE_FORMAT(o.timestamp, 'yyyy-MM') AS order_month,
        l.area AS area_name,
        COUNT(*) AS total_orders
    FROM orders_bronze o    
    JOIN locations_bronze l
            ON o.delivery_location_id = l.location_id
        GROUP BY 
           DATE_FORMAT(o.timestamp, 'yyyy-MM'),
           l.area

),

ranked_areas AS(
    SELECT *,
       DENSE_RANK() OVER(
        PARTITION BY order_month
        ORDER BY total_orders DESC
       )AS area_rank

    FROM monthly_orders   
)
SELECT 
     order_month,
     area_name,
     total_orders
FROM ranked_areas
WHERE area_rank <= 10
ORDER BY order_month, area_rank;    
```

q10

```python
%sql
With customer_revenue AS(
    SELECT 
         c.customer_id,
         YEAR(c.signup_date) AS signup_year,
         SUM(o.total_amount) AS lifetime_value FROM customers_filtered c
         JOIN orders_bronze o
         ON c.customer_id = o.customer_id
         GROUP BY c.customer_id,YEAR(c.signup_date)
)
SELECT signup_year,ROUND(AVG(lifetime_value),2) AS avg_customer_lifetime_value
FROM customer_revenue
GROUP BY signup_year
ORDER BY signup_year ;
```
